```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "timestamp": [
        "2026-08-25 10:00:01",
        "2026-08-25 10:00:03",
        "2026-08-25 10:00:05",
        "2026-08-25 10:00:07",
        "2026-08-25 10:00:09",
        "2026-08-25 10:00:11",
        "2026-08-25 10:00:13",
        "2026-08-25 10:00:15",
        "2026-08-25 10:00:17",
        "2026-08-25 10:00:19",
        "2026-08-25 10:00:21",
        "2026-08-25 10:00:23",
        "2026-08-25 10:00:25",
        "2026-08-25 10:00:27",
        "2026-08-25 10:00:29",
        "2026-08-25 10:00:31",
        "2026-08-25 10:00:33",
        "2026-08-25 10:00:35",
        "2026-08-25 10:00:37",
        "2026-08-25 10:00:39",
        "2026-08-25 10:00:41",
        "2026-08-25 10:00:43",
        "2026-08-25 10:00:45",
        "2026-08-25 10:00:47",
        "2026-08-25 10:00:49"
    ],
    "ip_address": [
        "192.168.1.10",
        "192.168.1.20",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.50",
        "192.168.1.30",
        "192.168.1.40",
        "192.168.1.20",
        "192.168.1.10",
        "192.168.1.30"
    ]
}

df = pd.DataFrame(data)

print("Network Log Data:")
display(df)

request_count = df["ip_address"].value_counts()

print("\nRequest Count by IP Address:")
display(request_count.reset_index().rename(columns={"index": "IP Address", "ip_address": "Request Count"}))

threshold = 10

suspicious_ips = request_count[request_count > threshold]

print("\nPossible DoS Attack Detection:")

if len(suspicious_ips) > 0:
    for ip, count in suspicious_ips.items():
        print(f"ALERT: Possible DoS attack detected from {ip}")
        print(f"Total requests: {count}")
else:
    print("No possible DoS attack detected.")

plt.figure(figsize=(10, 5))
request_count.plot(kind="bar")
plt.title("Number of Requests from Each IP Address")
plt.xlabel("IP Address")
plt.ylabel("Number of Requests")
plt.xticks(rotation=45)
plt.show()
```
