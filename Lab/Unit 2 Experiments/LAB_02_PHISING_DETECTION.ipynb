```python
url = input("Enter URL: ")

phishing_words = ["login", "verify", "update", "bank", "secure"]

found = False

for word in phishing_words:
    if word in url.lower():
        found = True

if found:
    print("Possible Phishing URL")
else:
    print("Looks Safe")
```
